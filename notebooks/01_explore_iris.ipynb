## 1. Load the data

Each flower has four measurements in centimeters. The target is its species. We will keep all five columns together in a pandas DataFrame to make inspection easy.

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.datasets import load_iris

iris = load_iris(as_frame=True)
flowers = iris.data.copy()
flowers["species"] = iris.target.map(dict(enumerate(iris.target_names)))

print(f"Rows: {len(flowers)}")
print(f"Measurement columns: {list(iris.feature_names)}")
print(f"Species: {list(iris.target_names)}")
```

There are **150 rows**, one per flower. The four measurement columns are the **features**; `species` is the **target** we would eventually like a model to predict. The measurements describe sepal length, sepal width, petal length, and petal width.

## 2. Inspect a few rows

A row is one example. The measurements are numeric; the species is a category.

## 3. Check data types and missing values

Missing values can affect analysis and model training. This dataset is already clean, but checking is a useful habit.

```python
flowers.info()
```

```python
flowers.isna().sum()
```

The columns should have numeric measurement types and a text species type. The missing-value count should be zero for every column.

## 4. Count the species

A classification target is the category the model learns to predict. Check how many examples of each species are present.

```python
species_counts = flowers["species"].value_counts().sort_index()
species_counts
```

```python
species_counts.plot(kind="bar", color="steelblue", title="Number of flowers by species")
plt.xlabel("Species")
plt.ylabel("Number of flowers")
plt.xticks(rotation=0)
plt.show()
```

Each species has 50 examples, so the classes are balanced. This is useful context when we later evaluate a classifier.

## 5. Summarize the measurements

The table below gives common summary statistics for each measurement. For example, `mean` is the average, while `min` and `max` show the range in this dataset.

```python
flowers[iris.feature_names].describe().round(2)
```

## 6. Visualize two measurements

This scatter plot compares petal length and petal width. Each point is a flower; its color indicates its species. Look for clusters and overlap. A plot can reveal patterns, but it does not by itself prove a model will generalize to new flowers.

```python
for species, group in flowers.groupby("species"):
    plt.scatter(group["petal length (cm)"], group["petal width (cm)"], label=species, alpha=0.75)

plt.xlabel("Petal length (cm)")
plt.ylabel("Petal width (cm)")
plt.title("Iris flowers: petal measurements by species")
plt.legend()
plt.show()
```
